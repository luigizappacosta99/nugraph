```python
import pynuml
import nugraph as ng
```

```python
def test_process_uboone():
    f = pynuml.io.File("/sps/lbno/zappacosta/v10_11/nugraph/pynuml/tests/MyFile_r20000001_s00000_ts583492.h5")
    processor = pynuml.process.HitGraphProducer(
        file=f,
        semantic_labeller=pynuml.labels.StandardLabels(),
        event_labeller=pynuml.labels.FlavorLabels(),
        label_vertex=False)
    plot = pynuml.plot.GraphPlot(
        planes=["u", "v", "y"],
        classes=pynuml.labels.StandardLabels().labels[:-1])
    f.read_data_all(use_seq_cnt=False, evt_part=0)
    evts = f.build_evt(0)
    i = 0
    print(len(evts))
    for evt in evts:
        _, data = processor(evt)
        if not data:
            i=i+1
            continue
        plot.plot(data, i, target='semantic', how='true', filter='show')
        plot.plot(data, i, target='instance', how='true', filter='true')
        i=i+1
        print(i)
    
    print("Sei sul cluster mona, che te plotti")

    return plot
```

```python
test_process_uboone()
```
